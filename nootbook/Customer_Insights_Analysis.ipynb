<h1>Step 6: Run Hypothesis Tests</h1>

<h3><span style="color: #800000;">Test 1: Independent t-test (Gender vs Monthly Spend)</span></h3>
<h3>Null Hypothesis (H₀)</h3>
<p class="isSelectedEnd">Male and female customers have the same average monthly spending.</p>

<h3>Alternative Hypothesis (H₁)</h3>
Male and female customers have different average monthly spending.

```python
from scipy.stats import ttest_ind

# Create groups and remove missing values
male = df.loc[df["Gender"] == "Male", "MonthlySpend"].dropna()
female = df.loc[df["Gender"] == "Female", "MonthlySpend"].dropna()

# Perform Independent t-test
t_stat, p_value = ttest_ind(male, female)

# Display results
print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_value:.4f}")

# Decision
alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis (significant difference in Monthly Spend).")
else:
    print("Fail to Reject the Null Hypothesis (no significant difference in Monthly Spend).")
```

<h3><span style="color: #008080;">Test 2: One-Way ANOVA (Education vs Monthly Spend)</span></h3>
<h3>Null Hypothesis (H₀)</h3>
<p class="isSelectedEnd">The average monthly spending is equal across all education levels.</p>

<h3>Alternative Hypothesis (H₁)</h3>
At least one education level has a different average monthly spending.

```python
from scipy.stats import f_oneway

# Create groups based on Education
groups = [
    group["MonthlySpend"].dropna().values
    for name, group in df.groupby("Education")
]

# Perform One-Way ANOVA
f_stat, p_value = f_oneway(*groups)

# Display results
print("F-statistic:", f_stat)
print("P-value:", p_value)

# Decision
if p_value < 0.05:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

<h3><span style="color: #000080;">Test 3: Chi-Square Test (Married vs Number of Pets)</span></h3>
<h3>Null Hypothesis (H₀)</h3>
<p class="isSelectedEnd">Marital status and number of pets are independent.</p>

<h3>Alternative Hypothesis (H₁)</h3>
Marital status and number of pets are associated.

```python

from scipy.stats import chi2_contingency
# Create contingency table
table = pd.crosstab(
    df["Married"],
    df["NumPets"]
)

# Perform Chi-Square Test
chi2, p_value, dof, expected = chi2_contingency(table)

# Display results
print("Chi-square Statistic:", chi2)
print("Degrees of Freedom:", dof)
print("P-value:", p_value)

# Decision
if p_value < 0.05:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

<h3><span style="color: #008000;">Test 4: Pearson Correlation (Age vs DaysSinceLastInteraction)</span></h3>
<h3>Null Hypothesis (H₀)</h3>
<p class="isSelectedEnd">There is no significant relationship between Age and DaysSinceLastInteraction.</p>

<h3>Alternative Hypothesis (H₁)</h3>
There is a significant relationship between Age and DaysSinceLastInteraction.

```python
from scipy.stats import pearsonr

# Select relevant columns and remove missing values
data = df[["Age", "DaysSinceLastInteraction"]].dropna()

# Calculate Pearson correlation
corr, p_value = pearsonr(
    data["Age"],
    data["DaysSinceLastInteraction"]
)

# Display results
print(f"Correlation Coefficient: {corr:.4f}")
print(f"P-value: {p_value:.4f}")

# Significance level
alpha = 0.05

# Decision
if p_value < alpha:
    print("Reject the Null Hypothesis")
    print("There is a statistically significant correlation between Age and DaysSinceLastInteraction.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("There is no statistically significant correlation between Age and DaysSinceLastInteraction.")
```

<h3><span style="color: #ff0000;">Test 5: One-Way ANOVA (State vs Monthly Spend)</span></h3>
<h3>Null Hypothesis (H₀)</h3>
<p class="isSelectedEnd">Average monthly spending is equal across all states.</p>

<h3>Alternative Hypothesis (H₁)</h3>
At least one state has a different average monthly spending.

```python
from scipy.stats import f_oneway

# Create groups based on State
state_groups = [
    group["MonthlySpend"].dropna().values
    for _, group in df.groupby("State")
]

# Perform One-Way ANOVA
f_stat, p_value = f_oneway(*state_groups)

# Display results
print(f"F-statistic: {f_stat:.4f}")
print(f"P-value: {p_value:.4f}")

# Significance level
alpha = 0.05

# Decision
if p_value < alpha:
    print("Reject the Null Hypothesis")
    print("There is a significant difference in average Monthly Spend among different states.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("There is no significant difference in average Monthly Spend among different states.")
```

<h4><span style="color: #800000;">Check Assumption 1: Normality Test (Shapiro-Wilk)</span></h4>

```python
from scipy.stats import shapiro

# Remove missing values
monthly_spend = df["MonthlySpend"].dropna()

# Perform Shapiro-Wilk test
stat, p_value = shapiro(monthly_spend)

# Print results
print(f"Shapiro-Wilk Statistic: {stat:.4f}")
print(f"P-value: {p_value:.4f}")

# Significance level
alpha = 0.05

# Interpretation
if p_value > alpha:
    print("Fail to Reject the Null Hypothesis")
    print("The data appears to follow a normal distribution.")
else:
    print("Reject the Null Hypothesis")
    print("The data does not appear to follow a normal distribution.")
```

<h4><span style="color: #000080;">Check Assumption 2: Homogeneity of Variance (Levene Test)</span></h4>

```python
from scipy.stats import levene

# Create groups and remove missing values
male = df.loc[df["Gender"] == "Male", "MonthlySpend"].dropna()
female = df.loc[df["Gender"] == "Female", "MonthlySpend"].dropna()

# Perform Levene's Test
stat, p_value = levene(male, female)

# Print results
print(f"Levene Statistic: {stat:.4f}")
print(f"P-value: {p_value:.4f}")

# Significance level
alpha = 0.05

# Interpretation
if p_value > alpha:
    print("Fail to Reject the Null Hypothesis")
    print("The variances of the two groups are equal (homogeneity of variance is satisfied).")
else:
    print("Reject the Null Hypothesis")
    print("The variances of the two groups are significantly different (homogeneity of variance is violated).")
```

<h2><span style="color: #808000;">Interpretation of p-value</span></h2>
<ul data-spread="true">
 	<li><strong>If p-value &lt; 0.05:</strong> Reject the Null Hypothesis (H₀). There is a statistically significant difference or relationship.</li>
 	<li><strong>If p-value ≥ 0.05:</strong> Fail to Reject the Null Hypothesis (H₀). There is no statistically significant evidence to support the alternative hypothesis.</li>
</ul>

```python
# Remove missing values
data = df[["Age", "DaysSinceLastInteraction"]].dropna()

# Calculate correlation
correlation = data["Age"].corr(data["DaysSinceLastInteraction"])

# Display the result
print(f"Correlation Coefficient: {correlation:.4f}")

# Interpretation
if correlation > 0:
    print("There is a positive correlation between Age and DaysSinceLastInteraction.")
elif correlation < 0:
    print("There is a negative correlation between Age and DaysSinceLastInteraction.")
else:
    print("There is no correlation between Age and DaysSinceLastInteraction.")
```
